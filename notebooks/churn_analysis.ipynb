## Preprocessing Setup

```python
numeric_features = X_train.select_dtypes(include="number").columns
categorical_features = X_train.select_dtypes(include="object").columns
```

```python
print("Numerical:", list(numeric_features))
print("Categorical:", list(categorical_features))
```

## Numerical Preprocessing Pipeline

Ab hum numerical columns ke liye preprocessing banayenge.

Numerical columns:

SeniorCitizen
tenure
MonthlyCharges
TotalCharges

Hum 2 kaam karenge:

Numerical Data
      ↓
1. Missing values handle
      ↓
2. Scaling

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
```

```python
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

## Categorical Preprocessing

Ab hum categorical columns ko preprocess karenge.

Categorical columns mein text values hoti hain, jaise:

gender → Male / Female
Contract → Month-to-month / One year / Two year
InternetService → DSL / Fiber optic / No

ML model directly text ko understand nahi karta, isliye humein inko numbers mein convert karna hai.

```python
from sklearn.preprocessing import OneHotEncoder

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
```

## Dono Preprocessing Pipelines Ko Combine Karna

Ab hamare paas 2 separate pipelines hain:

numeric_pipeline → numerical columns ke liye
categorical_pipeline → categorical columns ke liye

Ab in dono ko ek single preprocessing object mein combine karenge.

```python
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])
```

```python
preprocessor.fit_transform(X_train)
```

## Training aur Testing Data ko Preprocess Karna

Ab preprocessor ready hai. Ab isko actual data par apply karenge.

```python
X_train_processed = preprocessor.fit_transform(X_train)

print("Training data shape:", X_train_processed.shape)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)

print("Training data shape:", X_train_processed.shape)
```

```python
X_test_processed = preprocessor.transform(X_test)

print("Testing data shape:", X_test_processed.shape)
```

## Pehla ML Model: Logistic Regression

Ab preprocessing complete ho gayi hai. Ab hum ML model train karenge.

Beginner project ke liye pehle Logistic Regression use karte hain.

```python
from sklearn.linear_model import LogisticRegression
```

```python
## Model Train Karo
model = LogisticRegression(max_iter=1000)
```

```python
model.fit(X_train_processed, y_train)
```

## Model Se Predictions Lena

Model train ho gaya. Ab dekhte hain ke test data par model kya predict karta hai.

```python
## Churn Prediction
y_pred = model.predict(X_test_processed)

print(y_pred[:20])
```

```python
## Actual Values Dekho
print("Actual:   ", y_test.values[:20])
print("Predicted:", y_pred[:20])
```

## Model Evaluation: Confusion Matrix

Ab hum check karenge ke model ne kitni predictions correct aur incorrect ki hain.

```python
from sklearn.metrics import confusion_matrix

# Confusion matrix actual aur predicted values ko compare karti hai
cm = confusion_matrix(y_test, y_pred)

# Confusion matrix show karna
print(cm)
```

## Classification Report

Ab confusion matrix ke baad hum precision, recall aur F1-score ek saath dekhenge.

Precision → Model ne jin customers ko churn kaha, unmein kitne actually churn hue.

Recall → Jo customers actually churn hue, unmein se model ne kitne identify kiye.

F1-score → Precision aur Recall ka combined measure.

```python
from sklearn.metrics import classification_report

# Actual aur predicted values ki performance report banana
report = classification_report(y_test, y_pred)

# Complete classification report display karna
print(report)
```

## ROC-AUC Calculate Karna

Ab hum model ki ROC-AUC performance check karenge.

Iske liye sirf 0/1 predictions nahi, balki model ki probability chahiye.

```python
# Har customer ke liye churn ki probability nikalna
y_prob = model.predict_proba(X_test_processed)[:, 1]

# Pehli 10 churn probabilities dekhna
print(y_prob[:10])
```

## ROC-AUC batata hai:

Model churn aur non-churn customers ko kitni achhi tarah separate kar raha hai.
ROC-AUC = 0.50
→ Random-level separation

ROC-AUC = 0.80
→ Model classes ko achhi tarah separate kar raha hai

ROC-AUC = 1.00
→ Perfect separation

```python
from sklearn.metrics import roc_auc_score

# Actual churn values aur churn probabilities se ROC-AUC calculate karna
roc_auc = roc_auc_score(y_test, y_prob)

# ROC-AUC result display karna
print("ROC-AUC:", roc_auc)
```

## PR-AUC Calculate Karna
Ab hum PR-AUC (Precision-Recall AUC) calculate karenge.

Ye hamare churn project mein useful hai kyunke Churn class (1) minority class hai.

PR-AUC = churn class ko identify karte waqt precision aur recall ka balance kitna achha hai.

Ye especially useful hota hai jab positive class relatively kam ho.

```python
from sklearn.metrics import average_precision_score

# Actual churn values aur churn probabilities se PR-AUC calculate karna
pr_auc = average_precision_score(y_test, y_prob)

# PR-AUC result display karna
print("PR-AUC:", pr_auc)
```

## Model Metrics Ek Jagah Save Karna

Ab hum apne important metrics ko ek jagah store karenge, taake baad mein easily compare kar saken.

```python
from sklearn.metrics import precision_score, recall_score, f1_score

# Model ki precision calculate karna
precision = precision_score(y_test, y_pred)

# Model ki recall calculate karna
recall = recall_score(y_test, y_pred)

# Model ka F1-score calculate karna
f1 = f1_score(y_test, y_pred)

# Saare important metrics ek dictionary mein store karna
metrics = {
    "ROC-AUC": roc_auc,
    "PR-AUC": pr_auc,
    "Precision": precision,
    "Recall": recall,
    "F1-Score": f1
}

# Metrics display karna
print("\nMetrics:")
for metric, value in metrics.items():
    print(f"{metric}: {value}")
```

## Model Ko Simple Baseline Se Compare Karna
Ab ek important project requirement complete karte hain:

Hamara actual ML model ek simple baseline se better perform karna chahiye.

Baseline ka matlab hai ek bahut simple model jo proper ML learning nahi karta, sirf basic rule follow karta hai.

```python
## Dummy Baseline

from sklearn.dummy import DummyClassifier

# Simple baseline model banana
# Ye majority class ko predict karega
baseline = DummyClassifier(strategy="most_frequent")

# Baseline ko training data par train karna
baseline.fit(X_train_processed, y_train)

# Test data par baseline predictions lena
baseline_pred = baseline.predict(X_test_processed)

# Baseline predictions dekhna
print(baseline_pred[:20])
```

## Baseline Ka ROC-AUC Check Karna

Ab baseline ki performance bhi calculate karenge, taake Logistic Regression ke saath fair comparison ho.

Baseline ne har customer ke liye 0 predict kiya hai, lekin ROC-AUC ke liye probabilities/scores chahiye.

```python
from sklearn.metrics import roc_auc_score

# Baseline ki predicted probability nikalna
baseline_prob = baseline.predict_proba(X_test_processed)[:, 1]

# Baseline ka ROC-AUC calculate karna
baseline_roc_auc = roc_auc_score(y_test, baseline_prob)

# Baseline ROC-AUC display karna
print("Baseline ROC-AUC:", baseline_roc_auc)
print("Logistic Regression ROC-AUC:", roc_auc)
```

## Baseline aur ML Model Ka Comparison

Ab hum dono models ke important metrics ek table mein dekhenge.

Baseline ke liye mainly ROC-AUC compare karenge, aur Logistic Regression ke metrics bhi saath show karenge.


Ye table kyun banaya?

Ab project mein ek clear comparison available hai:
